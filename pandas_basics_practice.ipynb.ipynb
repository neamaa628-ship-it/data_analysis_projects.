```python
import numpy as np
x = np.linspace (0,10,100)
y = np.sin(x)
plt.plot(x,y,'o-')
plt.title("sin wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

```python
x = np.linspace (0,10,10)
y = np.sin(x)
plt.plot(x,y,'o-')
plt.title("sin wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

```python
x = np.linspace(0,10,50)
y = np.sin(x)

plt.plot(x,y,linestyle = '-.',color="purple",lw=5)
plt.title("sin wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True,color="red",linestyle='--')
plt.xlim(2,8)
plt.show()
```

```python
x= np.linspace(0,10,50)
y=np.sin(x) +np.random.rand(50)
sizes = np.random.randint(50,500,50)
colors = np.linspace(0,10,50)
plt.title("scatter graph" , fontsize=12 , color="brown")
plt.scatter(x,y,c=colors,s=sizes,marker='o',alpha=0.8,cmap='viridis',edgecolors='black')
plt.ylabel("y value")
plt.xlabel("x value")
plt.colorbar (label=("colorbar"))
plt.show()
```

```python
np.random.seed(3)
scores = np.random.normal(loc=80,scale=10,size=100)
scores=np.clip(scores,0,100)
plt.Figure(figsize=(10,6))
plt.hist(scores,10,color="skyblue",edgecolor="black")
plt.title("distribution of student score",fontsize=16)
plt.xlabel("score")
plt.ylabel("student")
plt.grid(axis='y',linestyle="--",alpha=0.5)
plt.show()
```
