```python
def split_rows(df):
    rows = []

    for _, row in df.iterrows():

        row1 = row.copy()
        row2 = row.copy()

        columns = [
            'fuel',
            'air_pollution_score',
            'city_mpg',
            'hwy_mpg',
            'cmb_mpg',
            'greenhouse_gas_score'
        ]

        for col in columns:
            values = str(row[col]).split('/')
            row1[col] = values[0]
            row2[col] = values[1]

        rows.append(row1)
        rows.append(row2)

    return pd.DataFrame(rows)
```

Are SmartWay vehicles more fuel-efficient than non-SmartWay vehicles in 2018
